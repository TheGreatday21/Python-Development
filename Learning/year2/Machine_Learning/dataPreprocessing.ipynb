```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt



## iam going to be creating this dataframe from a dictionary makes things faster and easier
numRows = 20
students = {
    'Access_No':[ f"M24B38/0{num}" if num >= 12 else f"S24B38/0{num}" for num in range(numRows) ],
    'Gender':np.random.choice(['M','F'],size=numRows),
    'Age':np.random.randint(20,30,size=numRows),
    'Course':np.random.choice(['DSC3108', 'DSC3114', 'MTH3108', 'SYE3104'],size=numRows),
    'Program':np.random.choice(['Day','Night'],size=numRows),
    'Residency':np.random.choice(['Non Resident','Halls'],size=numRows)

}



studentData = pd.DataFrame(students)

studentData.tail()
```
