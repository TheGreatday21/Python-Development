#### The main aim of this script is to learn to create my own dataset using pandas

```python
import pandas as pd
```

- Creating an empty dataframe.

```python
##creating an empty dataframe
dataFrame = pd.DataFrame()
print(dataFrame)
```

- Creating a dataframe from a list

```python
##creating a simple list 
fruits = ['Mango','Banana','Kiwi','Apple','Star Fruit']

##now create the dataframe using the list
dataFrame1 = pd.DataFrame(fruits)
print(dataFrame1)
```

- Creating a dataframe from a list of lists


```python
##creating a nested list
nestedList = [
    ['Lisa',23,True],
    ['James',12,False],
    ['Wakooli',34,False],
    ['Johnson',43]
]

#creating the dataframe 
##the difference here is that we ad the columns as an argument where the first index in each list is under the first column name and the second index under the second column name 
dataFrame2 = pd.DataFrame(nestedList,columns=['Name','Age','Pissed_me_off'])
print(dataFrame2)
```

- Creating a dataframe from a dictionary of ndarray

```python
##create a dict
##here the keys are the columns for our dataset and the values are in n dimensional arrays(technically lists with the values for that specific column)
students = {
    'Name': ['Elijah','Greatday21','Joshua','Baraka'],
    'Fruits':['Mango','Banana','Kiwi','Apple']
}

##NOTE : All the arrays must be of the same length for this to work

##creating the dataframe
dataFrame3 = pd.DataFrame(students)
print(dataFrame3)
```

- Creating a dataframe by passing a list of dictionaries and row indexes

```python
##creating a list of dictionaries
data = [
    {'Name':'Elijah', 'Age':24,'Telephone No':+256787106109},
    {'Name':'Johnson', 'Age':14,'Telephone No':+256790324234},
    {'Name':'Mariach', 'Age':22,'Telephone No':+256745434223},
    {'Name':'Obote', 'Age':54,'Telephone No':+256781000453}
]

##creating the dataframe 
##here instead of letting it give you default indexes that usually start from 0 you can specify the index you want
dataFrame4 = pd.DataFrame(data,index=[1,2,3,4])
print(dataFrame4)
```

- Creating the dataframe using the zip function.

```python
##create a list to marge two lists 
Name = ['Elijah','Greatday21','Joshua','Baraka']
Fruits = ['Mango','Banana','Kiwi','Apple']

#this will make the two lists rows in the dataframe 
listOfRows = list(zip(Name,Fruits))#using the zip function here to join the two lists

dataFrame5 =  pd.DataFrame(listOfRows,columns=['Names','Fruits'])
print(dataFrame5)



```
